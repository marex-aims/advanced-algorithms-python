```python
word = "Exaucé Kambale Maruba"
counter = {}
word = word.lower()
for letter in word:
    if letter not in counter:
        
        counter[letter] = 1
    else:
        counter[letter] += 1
print(counter)
```

```python
words = "Exaucé Maruba"
print(words.get('a', 0))
counter = {}
for word in words:
    counter[word] = counter.get(word, 0) + 1
    print(counter)
```
