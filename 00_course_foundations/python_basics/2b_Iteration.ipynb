## Part 0 · Why do we need loops?

### The problem
Suppose we want to welcome the class 3 times:

````python
print('Welcome to AIMS!')
print('Welcome to AIMS!')
print('Welcome to AIMS!')
````

**Question:** what if we needed 1000 welcomes? Copy-paste 1000 lines? Of course not.

A **loop** lets us write the instruction **once** and tell Python to **repeat** it.
One pass through the repeated block is called an **iteration**.

Python has two kinds of loops:
* `while` → "keep going **as long as** something is true"
* `for` → "do this **for each** item"

## Part 1 · The `while` loop


### 1.1 `if` versus `while`

An `if` checks its condition **once**. A `while` checks its condition **again and again**,
and keeps repeating its block as long as the condition is `True`.

**Predict**: how many times will each cell print something?

````python
n = 1
if n <= 3:
    print('if says: n is', n)
````

````python
n = 1
while n <= 3:
    print('while says: n is', n)
    n = n + 1
````

### 1.2 The 3 ingredients of a `while` loop

```
1. START      n = 1              <- give the variable a starting value
2. CONDITION  while n <= 3:      <- keep going as long as this is True
3. UPDATE         n = n + 1      <- change the variable, so the loop can end!
```

**How Python reads it:**
1. Check the condition. If it is `False` → leave the loop.
2. If it is `True` → run the indented block, then go back to step 1.

### 1.3 Live coding: count from 1 to 5

We build it **line by line** (let's type it together):
1. `n = 1`: the start
2. `while n <= 5:`: the condition
3. `print(n)`: what we repeat
4. `n = n + 1`: the update

````python
n = 1
while n <= 5:
    print(n)
    n = n + 1
print('Done!')
````

### 1.4 Tracing: following the loop by hand

*Tracing* means acting like the computer, one pass at a time. This is the **most important skill** for understanding loops.

| Pass | `n` at the start | `n <= 5` ? | prints | `n` after the update |
|:---:|:---:|:---:|:---:|:---:|
| 1 | 1 | True | 1 | 2 |
| 2 | 2 | True | 2 | 3 |
| 3 | 3 | True | 3 | 4 |
| 4 | 4 | True | 4 | 5 |
| 5 | 5 | True | 5 | 6 |
| 6 | 6 | **False** | — | the loop stops |

Then Python continues with the first line **after** the loop: `print('Finished!')`.

### 1.5 A countdown

Same 3 ingredients, but now the update **subtracts**.

````python
n = 5

while n >= 1:
    print(n)
    n = n - 1
print('Done!')
````

### Recall: the remainder operator `%`
`a % b` gives the **remainder** of the division of `a` by `b`. We will use it quite often today.

````python
print(7 % 2)    # 7 = 3*2 + 1  -> remainder 1
print(8 % 2)    # 8 = 4*2 + 0  -> remainder 0
print(10 % 3)   # 10 = 3*3 + 1 -> remainder 1
````

So a number `n` is **even** when `n % 2 == 0`, and `n` is a **multiple of 3** when `n % 3 == 0`.

## Part 2 · Infinite loops

**Goal:** know what happens when the update is forgotten, and how to stop it.

If nothing inside the loop ever makes the condition `False`, the loop **never ends**. This is an **infinite loop**.

```python
n = 1
while n <= 5:
    print(n)       # we forgot n = n + 1  ->  prints 1 forever!
```

**How to stop it:** click the **Stop / Interrupt** button (■) in Jupyter or Colab (or *Kernel → Interrupt*).


````python
# This loop will run forever
# n = 1
# while n <= 5:
#     print(n)
````

**Checklist when a loop never stops:**

- did I write the **update**?
- Does the update move the variable **towards** making the condition `False`?

## Part 3 - Asking until the answer is valid, and `break`

**Goal:** use `while` when we **do not know in advance** how many repetitions are needed.

### 3.1 Try it: the password
We cannot know how many attempts the user will need, so `while` is the perfect tool.

````python
password = input('Enter the password: ')

while password != 'aims':
    print('Wrong password, try again.')
    password = input('Enter the password: ')
print('Welcome!')
````

**Pattern "ask until valid":**
1. Ask **once before** the loop (so the condition has something to test: this is the *start*).
2. Ask **again inside** the loop (this is the *update*).

### 3.2 `break`: leave the loop immediately

`break` stops the loop **right now**, wherever Python is inside it.
A very common pattern is `while True:` (a loop that is always "on") together with a `break` that turns it off.

````python
while True:
    word = input('Type a word (or "stop" to finish): ')
    if word == 'stop':
        break
    print('You typed:', word)
print('Bye!')
````

`while True:` **without** a `break` is an infinite loop. Always make sure a `break` can be reached.

## Part 4 · The `for` loop

**Goal:** go through the characters of a string, one at a time, with `for`.

With `while` we had to write the start, the condition and the update ourselves.
With `for`, Python does all that bookkeeping **for us**.

```
for item in sequence:
    ...do something with item...
```

* `for` starts the loop
* `item` is a **temporary variable** (we choose its name): at each pass it holds the **next** item
* `in sequence` says *where* the values come from
* the **colon `:`** and the **indentation** work exactly like with `if` and `while`


Let's try the following code:

````python
for letter in 'AIMS':
    print(letter)
print('Done!')
````

| Pass | `letter` | printed |
|:---:|:---:|:---:|
| 1 | `'A'` | A |
| 2 | `'I'` | I |
| 3 | `'M'` | M |
| 4 | `'S'` | S |

The loop stops by itself when the string has no more characters. No update, no risk of an infinite loop.

## Part 5 · `range()`: looping over numbers

**Goal:** use `for` with `range(...)` to repeat things a given number of times, or to loop over numbers.

`range()` produces a sequence of numbers for the `for` loop to use.

**Format:** `range(start, stop, step)`
* `start` is **included** (default `0`)
* `stop` is **never included**
* `step` is the jump between numbers (default `1`, can be negative)

| Call | Numbers produced |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(2, 11, 2)` | 2, 4, 6, 8, 10 |
| `range(10, 0, -1)` | 10, 9, 8, ..., 1 |



````python
for i in range(5):
    print(i)
````

**Predict** the output of the next cell before running it.

````python
for i in range(1, 6):
    print(i)
````

````python
for i in range(2, 11, 2):
    print(i)
````

````python
for i in range(10, 0, -1):
    print(i)
print('Done!')
````

### Repeating something *n* times
When we don't care about the number itself, `range(n)` is the easiest way to repeat something exactly `n` times:

````python
for i in range(3):
    print('Welcome to AIMS!')
````

(Remember Part 0? Now 1000 welcomes is just `range(1000)`.)

### The off-by-one trap
`range(5)` gives **5** numbers, but they are `0` to `4`. The `stop` value is never included.
If you want `1` to `5`, write `range(1, 6)`.

**Predict** (write your answer on paper, then run): what does this print?

````python
for i in range(3, 8):
    print(i)
````

## Part 6 · Counting and summing

Imagine a **scoreboard** that starts at `0`. Each time something happens, we update the scoreboard.

```
1. BEFORE the loop:   total = 0          <- create the scoreboard
2. INSIDE the loop:   total = total + ...  <- update it
3. AFTER the loop:    print(total)       <- read the result
```

### 6.1 Summing: add up 1 + 2 + 3 + 4 + 5

````python
total = 0

for i in range(1, 6):
    total = total + i
print('Sum:', total)
````

**Trace** (what is `total` after each pass?):

| Pass | `i` | `total` before | `total` after |
|:---:|:---:|:---:|:---:|
| 1 | 1 | 0 | 1 |
| 2 | 2 | 1 | 3 |
| 3 | 3 | 3 | 6 |
| 4 | 4 | 6 | 10 |
| 5 | 5 | 10 | **15** |

**Two classic mistakes**
1. Forgetting `total = 0` **before** the loop → Python complains that `total` does not exist.
2. **Indenting the final `print`** → it prints after *every* pass instead of once at the end.

````python
# Mistake 2: the print is indented, so it is INSIDE the loop
total = 0
for i in range(1, 6):
    total = total + i
    print('Sum so far:', total)
````

### 6.2 Counting: how many times does something happen?
Same idea, but the scoreboard goes up by **1**, and only when a condition is true (so we put an `if` inside the loop).

````python
word = 'banana'
count = 0
for letter in word:
    if letter == 'a':
        count = count + 1
print('The letter a appears', count, 'times')
````

## Part 7 · `for` or `while`?

| Use `for` when... | Use `while` when... |
|---|---|
| you know **what to go through** (letters, numbers) | you only know **when to stop** |
| *"for each letter in the word"* | *"until the user types the right password"* |
| *"repeat 10 times"* | *"as long as the number is positive"* |

Tip:
- If you can say **"for each ..."** or **"repeat n times"**, use `for`.
- If you say **"until ..."** or **"as long as ..."**, use `while`.

## Part 9 · Wrap-up

```python
# while: repeat as long as the condition is True
n = 1                      # 1. start
while n <= 5:              # 2. condition
    print(n)
    n = n + 1              # 3. update  (never forget it!)

# for over a string
for letter in 'AIMS':
    print(letter)

# for over numbers: range(start, stop, step)   (stop is NOT included)
for i in range(1, 6):
    print(i)

# break: leave the loop immediately
while True:
    if ...:
        break

# summing / counting
total = 0
for i in range(1, 6):
    total = total + i
```
