## Intro: Why functions?

### The problem
We want to greet three students:



````python
print('Hello, Awa - welcome to AIMS Senegal!')
print('Hello, Moussa - welcome to AIMS Senegal!')
print('Hello, Fatou - welcome to AIMS Senegal!')
````

Only the **name** changes. If we want to change the wording (say, `Welcome` instead of `welcome`), we must edit **every** line.



A **function** is a **named block of code** that we write **once** and can **call** (run) as many times as we like. Think of a function as a **machine**:

```
   input(s)        ┌────────────┐        output
  ───────────────► │  FUNCTION  │ ───────────────►
   'AIMS'          │    type    │          str
                   └────────────┘
```

### You have *already* been using functions!
`print(...)`, `input(...)`, `int(...)`, `type(...)` are all functions.

````python
print(type('AIMS'))
print(int('42'))
````

### Vocabulary

| Word | Meaning | Example |
|---|---|---|
| **call** a function | run it | `type('AIMS')` |
| **argument** | the value we give to the function | `'AIMS'` |
| **return value** | the value the function gives back | `str` |

## Part 3 · Parameters: giving inputs to a function

**Goal:** make functions flexible by giving them inputs.

### 3.1 One parameter
Our `say_hello()` always says the same thing. We want to greet **any** name. So we add a **parameter**, a variable listed in the parentheses of the `def`.

````python
def greet(name):
    print('Hello,', name, '- welcome to AIMS Senegal!')

greet('Awa')
greet('Moussa')
greet('Fatou')
````

**What happens when we write `greet('Awa')`?**
1. Python **copies** the argument `'Awa'` into the parameter `name`.
2. It runs the body with `name = 'Awa'`.
3. The call is finished, `name` disappears.

| Word | Where | Example here |
|---|---|---|
| **parameter** | in the `def` line | `name` |
| **argument** | in the call | `'Awa'` |

The argument can also be a variable:

````python
student = 'Fatou'
greet(student)        # argument as a variable
````

### 3.2 Several parameters
Separate the parameters with commas. When we call the function, the arguments are matched to the parameters **by position** (first with first, second with second...).

````python
def introduce(name, country):
    print('Hi! I am', name, 'and I come from', country)

introduce('Awa', 'Senegal')
introduce('Kofi', 'Ghana')
````

**The order matters!** Predict what this prints:

````python
introduce('Senegal', 'Awa')
````

Python does not know that `'Senegal'` is not a name. It just follows the positions.

**The number of arguments must match too.** The call `introduce('Awa')` would crash with a `TypeError` (missing `country`).

````python
introduce('Awa')    # This will raise a TypeError
````

## Part 4 · `return` versus `print`

**Goal:** understand the single most important idea about functions.

### 4.1 The problem
All our functions so far **print** their result. Let's try to *use* that result in a calculation.

**Predict**, then run:

````python
def show_square(x):
    print(x ** 2)

#value = show_square(5)
print('value is:', show_square(5))
````

`25` appeared on the screen, but `value` is `None` (Python's word for *"nothing"*). The function **showed** the result but did **not give it back**.

**The machine analogy**
* `print` = the machine **shows** the result on a screen. You can *look* at it, but you cannot take it away.
* `return` = the machine **hands** the result to you. You can store it, reuse it, calculate with it.



### The same function with `return`
Let's adjust our function by replacing print with **return**

````python
def square(x):
    return x ** 2

value = square(5)
print('value is:', value)
print(square(3) + square(4))      # we can calculate with the results!
````

| | `print` | `return` |
|---|---|---|
| What it does | **shows** a value on the screen | **gives** a value back to the caller |
| Can we store / reuse the result? | No | Yes |
| Does the function stop? | No | **Yes, immediately** |
| If a function has no `return` | it gives back `None` | |

**Rule of thumb:** if you might need the result later → `return`. If you only want to display something → `print`.



### 4.3 `return` ends the function immediately
Anything written **after** a `return` is never executed.

````python
def test():
    return 'I stop here'
    print('This line is never reached')

print(test())
````

## Part 5 · Functions with `if` and loops

**Goal:** combine everything you know (conditionals + loops) inside functions.

### 5.1 Functions with `if`
A function can contain `if / elif / else`, and can `return` different values.

````python
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False

print(is_even(4))
print(is_even(7))
````

The comparison `n % 2 == 0` **is already** `True` or `False`, so we can shorten the function:

````python
def is_even(n):
    return n % 2 == 0

print(is_even(4))
print(is_even(7))
````

````python
def grading(score):
    if score >= 70:
        return 'Excellent'
    elif score >= 50:
        return 'Pass'
    else:
        return 'Fail'

print(grading(85))
print(grading(55))
print(grading(30))
````

### 5.2 Functions with a loop
Remember the **summing pattern** from the iteration notebook? Let's put it in a function.

````python
def sum_up_to(n):
    total = 0
    for i in range(1, n + 1):
        total = total + i
    return total

print(sum_up_to(5))      # 15
print(sum_up_to(20))     # 210  (same as last session!)
````

The `return` is **after** the loop (not indented inside it). If it were inside, the function would stop after the very first pass!

And the **counting pattern**:

````python
def count_letter(word, letter):
    count = 0
    for ch in word:
        if ch == letter:
            count = count + 1
    return count

print(count_letter('banana', 'a'))     # 3
````

## Part 6 · Local variables

**Goal:** understand that a function has its own private workspace.

Variables created **inside** a function are **local**: they exist only while the function is running, and are invisible outside.

````python
def add_tax(price):
    total = price * 1.18        # 'total' is local to add_tax
    return total

print(add_tax(100))
````

If you remove the `#` below and run. What does Python say?

````python
print(total)
````

**A function talks to the rest of the program in only two ways:**

1. **arguments go in** (parameters)
2. **results come out** (`return`).

Also, a parameter is a **copy**: changing it inside the function does not change the variable outside.

**Predict**, then run: what do the two `print` lines show?

````python
x = 5

def mystery(x):
    x = x + 100
    return x

print(mystery(x))
print(x)
````

## Part 7: Default values

**Goal:** make some arguments optional.

We give a parameter a **default value** with `=`. If the caller does not provide that argument, the default is used.

````python
def greet_student(name, greeting='Hello'):
    print(greeting, name)

greet_student('Awa')                    # uses the default 'Hello'
greet_student('Awa', 'Salut')           # overrides the default
greet_student('Awa', greeting='Hola')   # we can also name the argument
````

**Rule:** parameters **with** a default go **after** those without. `def f(a=1, b):` is a `SyntaxError`.

## Part 9 · Wrap-up

### Cheat sheet

```python
def function_name(parameter1, parameter2=default):   # def + name + (parameters) + :
    # body (indented)
    return result                                    # gives a value back

result = function_name(argument1, argument2)         # call it
```

| Idea | Remember |
|---|---|
| **Define vs call** | `def` only saves the recipe; the code runs when you **call** it |
| **Parameter vs argument** | parameter = name in the `def`; argument = value in the call |
| **`return` vs `print`** | `return` hands the value back; `print` only shows it |
| **No `return`** | the function gives back `None` |
| **Local variables** | exist only inside the function |
| **Defaults** | `def f(a, b=2)`; defaults come last |
