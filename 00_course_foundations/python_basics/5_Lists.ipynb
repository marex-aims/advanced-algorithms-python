## Part 0. What is a list?

Imagine we have the marks of 30 students. Creating 30 variables (`mark1`, `mark2`, ...) would be absurd. A **list** stores many values in **one** variable.

A list is a sequence of values called **elements** (or **items**), written between square brackets and separated by commas.

### Live coding

````python
marks = [12, 15, 9, 18, 7]
print(marks)
print(type(marks))
print(len(marks))
````

The elements can be of any type, even mixed:

````python
names = ['Awa', 'Moussa', 'Fatou']
mixed = [1, 'two', 3.0, True]
empty = []

print(names)
print(mixed)
print(empty)
print(len(empty))
````

**Key idea:** a string is a sequence of **characters**; a list is a sequence of **any values**. Most of what you learned about strings (indexing, slicing, `len`, `in`, `for`) works the same way for lists.

## Part 3. Looping over a list

### Using `for` over the elements

````python
cars = ['Toyota', 'BMW', 'Alfa Romeo']
for car in cars:
    print(car)
````

### The sum pattern, the count pattern
Same recipe as before: create the variable **before** the loop, update it **inside**, print it **after**.

````python
marks = [12, 8, 15, 10, 6]

total = 0
for mark in marks:
    total = total + mark
print('Sum:', total)

passed = 0
for mark in marks:
    if mark >= 10:
        passed = passed + 1
print('Passed:', passed)
````

### The maximum pattern
Start from the first element, and replace it whenever we meet a larger one.

````python
best = marks[0]
for mark in marks:
    if mark > best:
        best = mark
print('Best:', best)
````

### Changing elements: loop over the indices
To **modify** the list we need the positions, so we loop over `range(len(...))`.

````python
numbers = [10, 20, 30, 40]
for i in range(len(numbers)):
    numbers[i] = numbers[i] * 2
print(numbers)
````

**Warning:** in `for number in numbers: number = number * 2`, only the temporary variable `number` changes, not the list.

````python
numbers = [10, 20, 30, 40]
for number in numbers:
    number = number * 2
    print(number)
print(numbers)
````

## Part 4. Operators and slicing

### The operators `+`, `*` and `in`
`+` joins two lists, `*` repeats a list, `in` tests whether a value is in the list.

````python
a = [1, 2, 3]
b = [4, 5]
print(a + b)
print(a * 2)
print(2 in a)
print(7 in a)
````

A quick way to create a list of five zeros: `[0] * 5`. We may use it in the exercises.

````python
print([0] * 5)
````



### Slicing
Exactly as for strings: `list[start:stop]`, with `stop` excluded.

````python
numbers = [4, 8, 15, 16, 23, 42]
print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
print(numbers[-2:])
````

A slice is a **new list**: the original is not changed. A handy trick: the step `-1` reverses a list.

````python
print(numbers[::-1])
print(numbers)
````

## Part 5. List methods

A **method** is called with a dot: `my_list.append(x)`. List methods **change the list itself** (unlike string methods, which return a new string).

| Method | What it does |
|---|---|
| `lst.append(x)` | adds `x` at the end |
| `lst.insert(i, x)` | inserts `x` at position `i` |
| `lst.remove(x)` | removes the first `x` |
| `lst.pop()` | removes and returns the last element |





### Adding elements

````python
fruits = ['mango', 'banana']
fruits.append('apple')
print(fruits)

fruits.insert(0, 'cherry')
print(fruits)
````

### Removing elements

````python
fruits.remove('banana')
print(fruits)

last = fruits.pop()
print(last)
print(fruits)

first = fruits.pop(0)
print(first)
print(fruits)
````

**Warning:** `remove` gives a `ValueError` if the value is not in the list. Test with `in` first:

````python
fruits = ['mango', 'banana']
if 'kiwi' in fruits:
    fruits.remove('kiwi')
else:
    print('kiwi is not in the list')
````

### Sorting and counting

| Method | What it does |
|---|---|
| `lst.sort()` | sorts the list in place |
| `lst.count(x)` | number of times `x` appears |

- The `.sort()` method rearranges the items in a list (from smallest to largest).
- Adding `reverse=True` flips that order and arrange them from largest to smallest instead.

````python
numbers = [12, 17, 8, 2, 78, 17]
numbers.sort()
print(numbers)

numbers.sort(reverse=True)
print(numbers)

print(numbers.count(17))
````

`sort()` changes the list. To get a **new** sorted list and keep the original, use the function `sorted()`:

````python
numbers = [12, 17, 8, 2]
new_list = sorted(numbers)
print(numbers)
print(new_list)
````

````python
names = ["Fatou", "Aby", "Moussa", "Chris"]
names.sort(reverse=True)
print(names)
````

**Warning (very common):** methods like `append` and `sort` **return `None`**. Never write `my_list = my_list.sort()`.

````python
numbers = [3, 1, 2]
result = numbers.sort()
print(result)
print(numbers)
````

## Part 7. Building a list in a loop

The recipe is the same as for strings, but the accumulator starts as the **empty list** `[]` and we use `append`.

```
result = []
for ...:
    result.append(...)
```

### Examples:
- a) Getting the first squares
- b) Then, keeping only some elements (for instance the even ones) using an `if` condition inside the loop

````python
# a) Getting the first squares
squares = []
for i in range(1, 6):
    square = i ** 2
    squares.append(square)
print(squares)
````

````python
# b) Keeping only even squares

evens = []
for number in squares:
    if number % 2 == 0:
        evens.append(number)
print(evens)
````

### Removing duplicates
Add an element only if it is **not already** in the new list using `not in`.

````python
numbers = [2, 5, 2, 8, 5, 9, 2]
unique = []
for number in numbers:
    if number not in unique:
        unique.append(number)
print(unique)
````

### From `range` to a list
`list(range(...))` turns a range into a real list.

````python
print(list(range(5)))
print(list(range(1, 11)))
print(list(range(10, 0, -3)))
````

### Reading values from the user

````python
marks = []
for i in range(3):
    mark = int(input('Mark: '))
    marks.append(mark)
print(marks)
print('Average:', sum(marks) / len(marks))
````

## Part 8. Lists and strings

| Tool | What it does |
|---|---|
| `list(s)` | list of the characters of the string `s` |
| `s.split()` | list of the words of `s` (cut at spaces) |
| `s.split(sep)` | list of the pieces of `s` cut at `sep` |
| `sep.join(lst)` | one string made of the strings of `lst`, separated by `sep` |

Notice:
- For split(), if no parameter given, split() uses any whitespace character (space, newline, tab) as separator.
- Not confuse `.split()` vs `.split(' ')`
### Examples

````python
print(list('AIMS'))

text = 'Python is fun'
print(text)
words = text.split()
print(words)
print(words[0])
print(len(words))
````

````python
date = '2026-10-08'
parts = date.split('-')
print(parts)
````

`join` is the opposite of `split`. The separator comes **first**:

````python
words = ['Python', 'is', 'fun']
print(' '.join(words))
print('-'.join(words))
print(''.join(words))
````

**Warning:** `join` only works on a list of **strings**. For numbers, convert each one with `str()` first.

## Part 9. Two names, one list; lists and functions

### 9.1 Two names, one list
With a number, `b = a` copies the value. With a list, `b = a` gives the **same list** a second name.

### Live coding
**Predict:** what does `a` contain after `b[0] = 99`?

````python
a = [1, 2, 3]
b = a
b[0] = 99
print(a)
print(b)
````

`a` changed too: there is only **one** list with two names. To get an independent **copy**, use a full slice (or `list(a)`):

````python
a = [1, 2, 3]
b = a[:]
b[0] = 99
print(a)
print(b)
````

### 9.2 Lists and functions
A function can take a list as a parameter and `return` a number, a boolean or a **new list**.

````python
def average(numbers):
    return sum(numbers) / len(numbers)

print(average([12, 8, 15, 10, 6]))
````

````python
def doubled(numbers):
    result = []
    for number in numbers:
        result.append(2 * number)
    return result

marks = [1, 2, 3]
print(doubled(marks))
print(marks)
````

**Warning:** a function that receives a list can **change the original** list (same reason as above).

````python
def add_zero(numbers):
    numbers.append(0)

marks = [1, 2, 3]
add_zero(marks)
print(marks)
````

**Key idea:** prefer functions that **return a new list** and leave their parameter untouched. They are easier to understand and to test.

## Part 10. Lists of lists: matrices

The elements of a list can be lists. A **list of lists** is a natural way to store a table or a **matrix**: each inner list is a **row**.

- `len(matrix)` returns the number of rows in the matrix.


````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

print(matrix)
print(len(matrix))
````

Two indices: `matrix[row][column]`. First we take the row (a list), then an element of that row.

````python
print(matrix[1])
print(matrix[1][2])
print(matrix[2][0])
````

### Looping over a matrix
One loop for the rows, one inner loop for the columns.

````python
for row in matrix:
    print(row)
````

You can get the first column as follows

````python
for row in matrix:
    print(row[0])
````

To get the column as list, you have to create a new list and use `.append()`

````python
first_col = []
for row in matrix:
    first_col.append(row[2])
print(first_col)
````

One loop over the rows, one inner loop for the columns.

````python
total = 0
for row in matrix:
    for value in row:
        total = total + value
print('Sum of all entries:', total)
````

When we need the positions, loop over the indices. For example, the **diagonal** is made of the entries `matrix[i][i]`:

````python
n = len(matrix)
for i in range(n):
    print(matrix[i][i])
````

## Part 11. A shortcut: list comprehension

A **list comprehension** builds a list in one line:

`[expression for item in sequence]`

Read it as: "for each item in the sequence, compute the expression and put the result in a new list". It is exactly the pattern of Part 7.

### Live coding: step by step, loop first

````python
squares = []
for i in range(1, 6):
    squares.append(i ** 2)
print(squares)
````

The same thing in one line:

````python
squares = [i ** 2 for i in range(1, 6)]
print(squares)
````

We can add a condition at the end (it plays the role of the `if` in the loop):

````python
numbers = [4, 15, 8, 23, 16, 42]
evens = [n for n in numbers if n % 2 == 0]
print(evens)

names = ['Fatou', 'Awa', 'Moussa']
print([name.upper() for name in names])
````

**Key idea:** a list comprehension is only a shorter way to write a loop with `append`. If it feels hard to read, write the loop.

## Part 12. Summary and self-check

### Cheat sheet
```python
lst = [4, 8, 15, 16, 23, 42]

lst[0]   lst[-1]   lst[1:4]   lst[::-1]       # indexing, slicing
lst[2] = 99                                   # lists are mutable
len(lst)  sum(lst)  max(lst)  min(lst)        # built-in functions
8 in lst                                      # membership

lst.append(x)   lst.insert(i, x)              # add
lst.remove(x)   lst.pop()   lst.pop(i)        # remove
lst.sort()      sorted(lst)    lst.count(x)   # sort / count

result = []                                   # build a list
for x in lst:
    result.append(x * 2)

words = 'a b c'.split()    ' '.join(words)    # strings <-> lists
matrix = [[1, 2], [3, 4]]  matrix[1][0]       # list of lists
b = a[:]                                      # an independent copy
```

### Classic mistakes
| Mistake | Remedy |
|---|---|
| `lst = lst.sort()` or `lst = lst.append(x)` makes `lst` equal to `None` | these methods change the list and return `None`; do not assign them |
| `for x in lst: x = x * 2` does not change the list | loop over `range(len(lst))` and assign `lst[i]` |
| `b = a` does not copy the list | use `b = a[:]` |
| `lst[len(lst)]` gives an `IndexError` | the last index is `len(lst) - 1`, or use `lst[-1]` |
| `lst.remove(x)` gives a `ValueError` | check `x in lst` first |
| `' '.join([1, 2])` gives a `TypeError` | `join` needs strings; convert with `str()` |
| starting a product at `0` | start a product at `1`, a sum at `0` |
