## Exercise 3. Points in the plane

A point is a tuple `(x, y)`.

1. Write `distance(p, q)` that returns the distance between two points

### Recall:

For two points $p = (x_1, y_1)$ and $q = (x_2, y_2)$, the distance is


$$d(p,q) = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}$$

***Example***

Let `p = (1, 2)` and `q = (4, 2)`.

- The distance is $(\sqrt{3^2 + 0^2} = \sqrt{9} = 3)$.

So the distance is **3**.


2. Write `midpoint(p, q)` that returns the midpoint (as a tuple).

The midpoint is halfway between the two points. Find the average of the x-coordinates and the average of the y-coordinates:

$[
M = \left(\frac{x_1+x_2}{2}, \frac{y_1+y_2}{2}\right)
]$

***Example***

Let `p = (2, 4)` and `q = (6, 8)`.

- Midpoint x-coordinate: \((2 + 6) / 2 = 4\)
- Midpoint y-coordinate: \((4 + 8) / 2 = 6\)

The midpoint is **`(4, 6)`**.
