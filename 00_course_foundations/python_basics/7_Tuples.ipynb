## Part 0. Why tuples?

Some values naturally go together:
- a point in the plane: `x` and `y`
- a student: a name and a mark
- a date: day, month, year

A list could hold them. But a point should not change by accident, and we would like Python to protect it. A **tuple** is a list that is **locked**.

**Key idea:** a tuple is an ordered sequence of values that cannot be modified after creation.

---
## Part 1. Creating a tuple

A tuple is written with **parentheses** and **commas**.

````python
point = (4, 7)
print(point)
print(type(point))
````

A tuple can hold values of different types, and any number of them:

````python
student = ('Awa', 19, 'Dakar')
print(student)
print(len(student))
````

The empty tuple is `()`. Parentheses are optional (the commas make the tuple), but we always write them for clarity.

````python
empty = ()
print(empty, len(empty))

also_a_point = 4, 7
print(also_a_point, type(also_a_point))
````

````python
t1 = ('Dakar')
t2 = ('Dakar',)
print(type(t1))
print(type(t2))
````

**Warning:** `(5)` is just the number 5 in parentheses. To make a tuple with **one** element, add a **trailing comma**: `(5,)`.

### From a list or a string

`tuple(...)` converts a sequence into a tuple.

````python
print(tuple([1, 2, 3]))
print(tuple('Dakar'))
````

## Part 2. Reading a tuple

Everything you know about reading a list works on a tuple: indexes, negative indexes, slices, `len`, `in`, loops.



````python
countries = ('South Africa', 'Senegal', 'Ghana', 'Cameroon', 'Tanzania')

print(countries[0])
print(countries[-1])
print(countries[1:4])        # a slice of a tuple is a new tuple
````

````python
print(len(countries))
print('Ghana' in countries)
for country in countries:
    print(country)
````

## Part 3. Tuples are immutable

With a list, `lst[1] = 99` is allowed. With a tuple, it is not.

Remove the `#` and run the cell:

````python
t = (10, 20, 30)
# t[1] = 99         # wil raise a TypeError
````

You get a `TypeError: 'tuple' object does not support item assignment`.

A tuple has no `append`, no `remove`, no `sort`. If you need a different tuple, **build a new one**:

We can use the Python's id() function to see that a new tuple has a different memory address.

````python
t = (10, 20, 30)
print(t)
print("Memory: ", id(t))

t = (t[0], 99, t[2])         # a brand-new tuple, assigned to the same name
print(t)
print("Memory: ", id(t))

t = t + (40, 50)             # + glues two tuples together (new tuple again)
print(t)
print("Memory: ", id(t))
````

**Key idea:** changing the *name* `t` so that it points to a new tuple is allowed. Changing the tuple *itself* is not.

### Why locked?
- Safety: a point or a date should not be modified by mistake.
- A tuple can be used as a **dictionary key** (Part 9), a list cannot.

**Warning.** Only the tuple is locked, not what it contains. If a tuple contains a list, that list can still change.

````python
t = (1, 2, [3, 4])
t[2].append(5)
print(t)
````

## Part 4. Comparing tuples

Tuples are compared **element by element, from left to right**. The first position where they differ decides.

````python
print((2, 5) < (3, 1))          # 2 < 3, decided at once
print((2, 5) < (2, 9))          # first equal, then 5 < 9
print(('Alice', 25) < ('Bob', 15))   # 'Alice' < 'Bob'; the numbers are never compared
````

If one tuple is the beginning of the other, the **shorter** one is smaller:

````python
print((1, 2) < (1, 2, 0))
````

**Why do we care?** Because Python can **sort** a list of tuples with this rule. We use it in Parts 7 and 8.

## Part 5. Unpacking and swapping

**Unpacking** puts the values of a tuple into separate variables, in one line.

````python
point = (4, 7)
x, y = point
print(x)
print(y)
````

````python
name, age, city = ('Awa', 19, 'Dakar')
print(name, 'is', age, 'and lives in', city)
````

**Warning:** the number of variables must match the number of values.

````python
# a, b = (1, 2, 3)      # remove the # : ValueError: too many values to unpack
````

### The swap

With lists or numbers you would need a third variable.

````python
# Swap using numbers
a = 4
b = 7

print(a,b)

# 1. Save 'a' in a temporary variable
temp = a      # temp = 4

# 2. Overwrite 'a' with 'b'
a = b         # a = 7

# 3. Assign the saved 'temp' value to 'b'
b = temp      # b = 4

print(a,b)
````

With tuples it is one line: the right side `b, a` is a tuple built **first**, then unpacked into `a, b`.

````python
a = 4
b = 7
print(a,b)

a, b = b, a                # swap in one line
print(a, b)
````

### A mathematical use: Fibonacci

Each step replaces `(a, b)` by `(b, a + b)`. Both new values use the **old** values, which is exactly what tuple assignment does.



````python
a, b = 0, 1
for i in range(10):
    print(a, end=' ')
    a, b = b, a + b
````

**Notice:** `print(a, end=' ')` outputs the value of `a` followed by a single space instead of jumping to a new line.

### Unpacking in a `for` loop

When a list contains tuples, we can unpack each one directly in the `for` line.

````python
students = [('Awa', 18), ('Samba', 12), ('Diarra', 15)]

for name, mark in students:
    print(name, 'has', mark)
````

## Part 6. Functions returning several values

A function can `return` several values separated by commas. Python packs them in a **tuple**; the caller unpacks them.



````python
def min_max(numbers):
    return min(numbers), max(numbers)

smallest, largest = min_max([15, 8, 42, 16, 4, 23])
print(smallest, largest)
````

If we store the result in **one** variable, we simply get the tuple:

````python
result = min_max([15, 8, 42, 16, 4, 23])
print(result)
print(type(result))
print(result[0])
````

Another example: the quotient and remainder of an integer division.

````python
def divide(a, b):
    return a // b, a % b

q, r = divide(17, 5)
print(q, r)
print('17 =', 5, '*', q, '+', r)
````

## Part 7. Dictionaries and tuples: `items()`

You already used `d.items()` in a loop. Now we can see what it really gives: a series of **tuples** `(key, value)`.



````python
fruits = {'banana': 7, 'apple': 12, 'orange': 15, 'mango': 5}

for pair in fruits.items():
    print(pair)                       # each pair is a tuple

for fruit, quantity in fruits.items():
    print(fruit, '->', quantity)      # unpacked in the for line
````

`list(d.items())` gives a list of tuples, and a list can be sorted. Thanks to Part 4, Python knows how to sort tuples: by the **first** element, then the second in case of a tie.

````python
pairs = list(fruits.items())
print(pairs)

pairs.sort()
print(pairs)
````

The pairs are now sorted by **key** (alphabetical order of the fruits).

## Part 8. Sorting a dictionary by value



To sort a dictionary by **value**, put the value **first**: build tuples `(value, key)`.

````python
fruits = {'banana': 7, 'apple': 12, 'orange': 15, 'mango': 5}

flipped = []
for fruit, quantity in fruits.items():
    flipped.append((quantity, fruit))     # the value comes first

print(flipped)

flipped.sort()
print(flipped)                 # smallest quantity first
````

To sort the values in descending order, use `reverse=True`

````python
flipped.sort(reverse=True)
print(flipped)                 # largest quantity first
````

**Notice**
Sorting the pairs sorts by keys, you can use a similar method without flipping the keys, values.


### Application: the most frequent words

We combine the counting pattern (`get`) and the `(count, word)` trick.

````python
text = 'the cat and the dog and the bird'

counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
````

````python
ranking = []
for word, n in counts.items():
    ranking.append((n, word))

ranking.sort(reverse=True)
print(ranking)
print(ranking[:3])
````

When two words have the same count, the tuple comparison looks at the **second** element: the words are then in reverse alphabetical order. This is the tie-break, for free.

## Part 9. Tuples as dictionary keys

A dictionary key must be **immutable**. Tuples are, so a tuple can be a key. This is useful when one value is not enough to identify an item.

A phone book where first names are not unique.

````python
phone = {
    ('Samba', 'Diop'): '555-0192',
    ('Fatou', 'Diop'): '555-0143',
    ('Fatou', 'Fall'): '555-0188',
}

print(phone[('Samba', 'Diop')])
````

````python
phone[('Awa', 'Seck')] = '555-0177'              # add
phone[('Fatou', 'Diop')] = '555-9999'            # update

for (first, last), number in phone.items():
    print(first, last, ':', number)
````

**Warning:** a list cannot be a key.

## Part 10. Summary and self-check

### Cheat sheet
```python
t = (4, 7)                    # a tuple
t = (5,)                      # one element: the comma is needed
t[0]    t[-1]    t[0:2]       # read like a list
len(t)  3 in t                # same as lists
x, y = t                      # unpacking (same number of variables)
a, b = b, a                   # swap
return a, b                   # a function returns a tuple
for k, v in d.items(): ...    # unpack each (key, value)
l.append((v, k)); l.sort(reverse=True)    # sort by value
d[(i, j)] = value             # tuple as key
```

### List or tuple?
| | List | Tuple |
|---|---|---|
| Brackets | `[1, 2]` | `(1, 2)` |
| Can be modified | yes | no |
| Can be a dictionary key | no | yes |
| Typical use | a collection that grows | a fixed record: point, pair, result |

### Classic mistakes
| Mistake | Fix |
|---|---|
| `t = (5)` is not a tuple | `t = (5,)` |
| `t[0] = 1` gives a `TypeError` | build a new tuple, or use a list |
| `a, b = (1, 2, 3)` gives a `ValueError` | one variable per value |
| `d[[1, 2]] = 0` gives a `TypeError` | use a tuple key `d[(1, 2)]` |
| sorting `d.items()` and expecting the order of the values | build `(value, key)` tuples |
