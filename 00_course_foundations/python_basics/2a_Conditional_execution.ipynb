# Part 1: Comparisons and boolean expressions

Python can compare values.

A comparison produces one of two Boolean values: `True` or `False`.

Example: The operator == compares two operands and produces


*   **True** if they are equal
*   **False** otherwise




````python
# Comparing integers
print(5 == 5)   # True
print(5 == 6)   # False
````

````python
# Comparing strings
print("Hello" == "Hello")     # True
print("Hello" == "hello")     # False
````

### A few more comparison operators

- `>` means greater than.
- `<` means less than.
- `>=` means greater than or equal to.
- `<=` means less than or equal to.
- `!=` means not equal to.


````python
# Comparisons
x = 5
y = 3

print(x > y)   # True
print(x < y)  # False

print(x <= y)  # False
print(x >= y)  # True

print(x != y)  # True
````

## Logical operators

There are three main logical operators, used to combine or invert boolean expressions:

| Operator | Meaning | True when... |
|---|---|---|
| `and` | conjunction | **both** sides are `True` |
| `or`  | disjunction | **at least one** side is `True` |
| `not` | negation   | flips `True` ↔ `False` |

**Precedence:** `not` binds tighter than `and`, which binds tighter than `or`. When in doubt (or when it helps readability), use parentheses `()` to group conditions explicitly.

### Examples

````python
age = 20

print(age > 12 and age < 18)   # False
print(age > 12 or age < 18)    # False
print(not (age > 18))          # False
````

### Combining Operators

Use parentheses `()` to make the grouping explicit — this avoids relying on precedence rules, and makes your intention obvious to anyone reading the code.



````python
age = 20

# False AND False is False, but checks both conditions
print((age > 12 and age < 18) or age >= 65)   # False

# False OR (True AND True)                    # True
print(age < 12 or (age >= 18 and age <= 25))
````

## Part 2: The `if` statement: run code only when a condition is true

Read the structure carefully:

```python
if condition:
    instruction
```

- The condition must be followed by a colon `:`.
- The instruction is **indented** (usually four spaces).
- If the condition is `True`, the indented instruction runs.
- If the condition is `False`, the instruction is skipped.

### Pitfall 2: writing `=` instead of `==`

This is one of the most common beginner mistakes. `if x = 5:` is **not** a comparison, it looks like an assignment inside a condition, and Python refuses it outright with a `SyntaxError`:

````python
x = 5
if x = 5:
    print('x is 5')
````

## Part 3 `if` and `else`: choose between two outcomes

Sometimes we want a message in either case. Use `else` for what should happen when the `if` condition is false.

```python
if condition:
    instruction_if_true
else:
    instruction_if_false
```

Notice that `else` has a colon but **no condition**.

if-else-logic.png width="200" height="150"

The `%` operator gives the remainder after division. For now, use this fact: a number is even when `number % 2 == 0`.

````python
x = 7

if x % 2 == 0:
  print ('even')
else:
  print ('odd')
````

### Pitfall: `else` never takes a condition

`else` catches "everything else" — it cannot be followed by its own boolean expression. Writing `else x > 0:` is a `SyntaxError` (if you want to test another condition, you need `elif`, covered next):

````python
x = 10

if x % 2 == 0:
    print('x is even')
else x > 10:                 # Will raise an error
    print('this is invalid')
````

````python
age = 16

if age >= 18:
    print("Adult")
else:
    print("Not an adult")
````

**Try it:** change `age` to `18`, then `25`. Both branches cannot run in the same execution: Python chooses one.

## Part 4 `if`, `elif`, and `else`: choose between several outcomes

When there are more than two possibilities, use `elif` (short for "else if").

Python looks down the chain and executes only the first block that evaluates to true. Once it finds a true condition, it skips the rest of the chain completely.

```python
if first_condition:
    ...
elif second_condition:
    ...
else:
    ...
```

if-elif-else-logic.png width="200" height="150"

````python
x = -1
y = 3

if x < y:
  print('x is less than y')
elif x > y:
  print('x is greater than y')
else:
  print('x and y are equal')
````

There is no limit on the number of elif statements.
If there is an else clause, it has to be at the end, but there doesn't have to be one.


````python
score = 72

if score >= 90:
  grade = 'A'
elif score >= 80:
  grade = 'B'
elif score >= 70:
  grade = 'C'
else :
  grade = 'F'
print ('Grade :', grade )
````


**Important:** order matters. For example, if `score >= 70` came before `score >= 90`, a score of 95 would match the first condition and print `"C"`. Put the more restrictive threshold first.

## Part 5 Nested conditionals: an `if` inside another `if`

Sometimes we need to check a second condition **only when the first condition is true**. We can put one `if` statement inside another. This is called a *nested conditional*.

The inner `if` must be indented one level further than the outer `if`.

````python
age = 20
has_ticket = True

if age >= 18:
    if has_ticket:
        print("Welcome!")
````

**Read the code from the outside in:**

1. Is `age >= 18` true?
2. Only if it is true, Python checks `has_ticket`.
3. `"Welcome!"` appears only when both conditions are satisfied.



**Try it:**

- then change it back to `20` and set `has_ticket = False`.

### Nested `if/else`

We can also use `else` inside a nested conditional to explain the second outcome.

````python
age = 20
has_ticket = False

if age >= 18:
    if has_ticket:
        print("Welcome!")
    else:
        print("You need a ticket.")
else:
    print("You must be at least 18.")
````

**Notice**

The inner `else` belongs to the inner `if`, because they have the same indentation.
