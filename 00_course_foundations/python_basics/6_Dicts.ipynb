## Part 0. Why dictionaries?

We store the ages of three students. With two lists, finding the age of Fatou requires a search:

### Example

````python
names = ['Awa', 'Moussa', 'Fatou']
ages = [23, 19, 21]

for i in range(len(names)):
    if names[i] == 'Fatou':
        print(ages[i])
````

With a **dictionary**, we simply ask for Fatou:

````python
ages = {'Awa': 23, 'Moussa': 19, 'Fatou': 21}
print(ages['Fatou'])
````

A dictionary (type `dict`) is a collection of **key: value** pairs, written between curly braces `{}`.

| Word | Meaning | Example |
|---|---|---|
| **key** | what we look up (unique) | `'Fatou'` |
| **value** | what we get back | `21` |
| **item** | one key-value pair | `'Fatou': 21` |

Think of a paper dictionary or a phone book: we look up a **word** (the key) to find its **definition** (the value), never the opposite.



Now we can print some information about our dictionary.

````python
print(ages)
print(type(ages))
print(len(ages))          # number of items

````

### Creating a dictionary
- When you write `ages = {'Awa': 23, 'Moussa': 19, 'Fatou': 21}`, you create a dictionary and immediately give it content.
- If you want to create an empty dictionary, you have two options:
```python
empty = dict()
empty = {}
```

````python
empty = {}                # an empty dictionary
print(empty)
print(type(empty))
print(len(empty))
````

## Part 2. The `in` operator: it tests the keys

`k in d` is `True` when `k` is a **key** of `d`. It is the safe way to avoid a `KeyError`.



````python
en2fr = {'one': 'un', 'two': 'deux', 'three': 'trois'}
print('one' in en2fr)
print('four' in en2fr)

word = 'four'
if word in en2fr:
    print(en2fr[word])
else:
    print('Unknown word')
````

**Predict:** is `'un'` in `en2fr`?

````python
print('un' in en2fr)
````

**Warning:** `'un'` is a **value**, not a key, so the answer is `False`. To test the values, use `d.values()`:

````python
print('un' in en2fr.values())
print('four' in en2fr.keys())
````

## Part 3. Dictionaries are mutable

Like lists, a dictionary can be changed in place.

### Live coding: change, add
Assigning to a key **changes** the value if the key exists, and **adds** a new item if it does not.

````python
student = {'name': 'Awa', 'age': 23, 'score': 14.5}

student['age'] = 24            # change
print(student)

student['city'] = 'Dakar'      # add
print(student)
````

### Deleting
`d.pop(key)` removes the item **and returns its value**. `del d[key]` removes it without returning anything.

````python
removed = student.pop('age')
print(removed)
print(student)

del student['score']
print(student)
````

### Building a dictionary in a loop
Same recipe as for lists: start with an empty dictionary, add items inside the loop.

````python
squares = {}
for i in range(1, 6):
    squares[i] = i ** 2
print(squares)
````

## Part 4. Looping over a dictionary

| Loop | Gives |
|---|---|
| `for key in d:` (or `d.keys()`) | the keys |
| `for value in d.values():` | the values |
| `for key, value in d.items():` | both, pair by pair |

### Live coding

````python
fruits = {'banana': 7, 'apple': 12, 'orange': 15, 'mango': 5}

print('Keys:')
for key in fruits:
    print(' ', key)

print('Values:')
for value in fruits.values():
    print(' ', value)

print('Pairs:')
for name, quantity in fruits.items():
    print(' ', name, '->', quantity)
````

**Warning:** looping directly over a dictionary gives only the **keys**. This is why `print(key, quantity)` fails with a `NameError`; use `fruits[key]` or `.items()`.

### The usual patterns, now with a dictionary
**Sum** of the values:

````python
total = 0
for quantity in fruits.values():
    total = total + quantity
print(total)
print(sum(fruits.values()))
````

**Filter**: keep the items satisfying a condition.

````python
for name, quantity in fruits.items():
    if quantity > 10:
        print(name, quantity)
````

**Maximum**: which key has the largest value? Same idea as the maximum of a list, but we remember the key too.

````python
best_name = None
best_quantity = 0
for name, quantity in fruits.items():
    if quantity > best_quantity:
        best_quantity = quantity
        best_name = name
print(best_name, best_quantity)
````

## Part 5. Counting with a dictionary

A very common use: **how many times does each thing occur?** The keys are the things, the values are the counts.

Plan for the letters of `'banana'` (expected: `b: 1, a: 3, n: 2`):
1. Start with an empty dictionary `counter`.
2. For each letter: if it is **not yet a key**, add it with count `1`; otherwise **add 1** to its count.



````python
word = 'banana'
counter = {}

for letter in word:
    if letter not in counter:
        counter[letter] = 1
    else:
        counter[letter] = counter[letter] + 1

print(counter)
````

Trace of the first passes:

| Pass | `letter` | `counter` after the pass |
|:---:|:---:|---|
| 1 | `'b'` | `{'b': 1}` |
| 2 | `'a'` | `{'b': 1, 'a': 1}` |
| 3 | `'n'` | `{'b': 1, 'a': 1, 'n': 1}` |
| 4 | `'a'` | `{'b': 1, 'a': 2, 'n': 1}` |

## Part 6. The `get()` method

`d.get(key, default)` returns `d[key]` if the key exists, and `default` otherwise, **never** a `KeyError`. Without `default`, it returns `None`.

### Live coding

````python
freq = {'b': 1, 'a': 3, 'n': 2}
print(freq.get('a', 0))
print(freq.get('c', 0))
print(freq.get('c'))
````

### The counter in one line
For a new letter, `get(letter, 0)` gives `0`, so we store `0 + 1`. For a known letter, it gives the current count, so we add 1 to it. No `if` needed.

````python
word = 'banana'
counter = {}
for letter in word:
    counter[letter] = counter.get(letter, 0) + 1
print(counter)
````

**Key idea:** the pattern `d[key] = d.get(key, 0) + 1` is the standard way to count with a dictionary.

### Counting words
`split()` turns a sentence into a list of words; then we count them exactly as we counted letters.

````python
sentence = 'the cat and the dog and the bird'
words = sentence.split()

word_count = {}
for word in words:
    word_count[word] = word_count.get(word, 0) + 1
print(word_count)
````

## Part 7. Displaying in a chosen order

A dictionary remembers the order of **insertion**, but it is **not sorted**. To display it in a chosen order, we sort the keys ourselves.

`sorted(d)` returns a **list** of the keys in increasing order.

### Live coding

````python
fruits = {'banana': 7, 'apple': 12, 'orange': 15, 'mango': 5}

print(sorted(fruits))
for name in sorted(fruits):
    print(name, fruits[name])
````

Reverse order, and sorting of the values:

````python
print(sorted(fruits, reverse=True))
print(sorted(fruits.values()))
````

**Key idea:** `sorted(d)` sorts the **keys**. The dictionary itself is not changed.

## Part 8. Dictionaries and lists together

Values can be **lists**. This lets us store several values under one key.

### Live coding: a list as a value

````python
grades = {'Awa': [12, 15, 9], 'Moussa': [8, 10], 'Fatou': [17, 14, 16]}

print(grades['Awa'])
print(grades['Awa'][0])
grades['Moussa'].append(12)
print(grades['Moussa'])
````

Average of each student: loop over the pairs, and apply `sum` and `len` to the list.

````python
for name, marks in grades.items():
    print(name, sum(marks) / len(marks))
````


### Records: a list of dictionaries
A list of dictionaries is a natural way to store a table: each dictionary is one row.

````python
students = [
    {'name': 'Awa', 'score': 14.5},
    {'name': 'Moussa', 'score': 9.0},
    {'name': 'Fatou', 'score': 17.0},
]

total = 0
for student in students:
    total = total + student['score']
print(total / len(students))

for student in students:
    if student['score'] >= 10:
        print(student['name'])
````

## Part 9. A mathematical example: polynomials

A polynomial such as $2x^3 - 5x + 7$ has many zero coefficients when its degree is large (think of $x^{1000} + 1$). We can store **only the non-zero terms** in a dictionary: the keys are the **exponents**, the values are the **coefficients**.

`{3: 2, 1: -5, 0: 7}` means $2x^3 - 5x^1 + 7x^0$.

### Live coding: evaluate the polynomial at a point

````python
p = {3: 2, 1: -5, 0: 7}
x = 2

value = 0
for exponent, coefficient in p.items():
    value = value + coefficient * x ** exponent
print(value)
````

## Part 10. Summary and self-check

### Cheat sheet
```python
d = {'a': 1, 'b': 2}          # create        {} is the empty dictionary

d['a']                        # read a value (KeyError if the key is missing)
d.get('z', 0)                 # read, with a default (no error)
'a' in d                      # is it a KEY?   ('a' in d.values() for values)
len(d)                        # number of items

d['c'] = 3                    # add or change
d.pop('a')                    # remove and return the value
del d['b']                    # remove

for k in d: ...               # keys
for v in d.values(): ...      # values
for k, v in d.items(): ...    # pairs

counter[x] = counter.get(x, 0) + 1     # counting pattern
sorted(d)                              # sorted list of the keys
groups[k] = []  then  groups[k].append(v)   # grouping
```

### Classic mistakes
| Mistake | Remedy |
|---|---|
| `d['missing']` gives a `KeyError` | test `key in d`, or use `d.get(key, default)` |
| `'Dakar' in capitals` is `False` (it is a value) | `in` tests keys; use `in capitals.values()` for values |
| `for x in d:` gives keys, not values or pairs | use `d.values()` or `d.items()` |
| expecting a dictionary to be sorted | it keeps insertion order; use `sorted(d)` |
| `d[k].append(x)` when `k` is missing | create `d[k] = []` first |
| a list as a key: `{[1, 2]: 'x'}` gives a `TypeError` | keys must be immutable (numbers, strings) |
| `{}` is not an empty set | `{}` is an empty dictionary |
