# Part 3: Variables

## Part 4: Expressions
### Left-hand side vs. Right-hand side

In Python assignment (`=`), code always runs **from right to left**:

* **Right-Hand Side (RHS):** The calculation or value evaluated *first* (called **expression**).
* **Left-Hand Side (LHS):** The variable name that receives and stores the result.

````python
age = 20 + 3
print(age)
````

### Updating a value using its current value


**Python calculates the RHS first, then stores the result in the LHS.**

````python
score = 10
score = score + 5

print(score)
````

** Notice **

Read `score = score + 5` as: "take the current score, add 5, and store the result back in `score`."

Through this example, you see the difference between the `=` assignment operator and mathematical equality (denoted as `==`).

`score = score + 5` would look strange as a math equation.

### Operator precedence

Python evaluates operators in a fixed order (from highest to lowest priority):

1. Parentheses `()` — always evaluated first, and can be used to force the order you want.
2. Exponentiation `**`
3. Multiplication `*`, division `/`, floor division `//`, modulo `%` — left to right.
4. Addition `+` and subtraction `-` (left to right).

**Example:** `1 + 2 ** 3 / 4 * 5`

- No parentheses.
- Exponentiation first: `2 ** 3` → `8`, giving `1 + 8 / 4 * 5`.
- `*` and `/` next, left to right: `8 / 4` → `2.0`, then `2.0 * 5` → `10.0`, giving `1 + 10.0`.
- Addition last: `1 + 10.0` → `11.0`.







````python
print(1 + 2 ** 3 / 4 * 5)   # 11.0, following precedence rules above
print((1 + 2) ** ((3 / 4) * 5))   # very different result once we force our own order with ()
````

# Part 7: Text is a string

Text values have the type `str`, short for **string**. Text must be enclosed in quotation marks.

````python
first_name = "Exaucé"
city = "Maruba"

print(first_name)
print(city)
print(type(first_name))
````

### Joining text with `+`

For strings, `+` joins pieces of text together. This is called **concatenation**.

````python
first_name = "Exaucé"
last_name = "Maruba"

full_name = first_name + " " + last_name
print(full_name)
````

**Notice**

The `" "` in the middle is a string containing one space. Without it, the names would be joined without a space.

Python knows the difference between numbers and text, and the `+` operator behaves differently depending on the types involved:

- `+` between two numbers means **addition**.
- `+` between two strings means **concatenation** (joining them together).



````python
u = 5 + 2
print(u)
````

````python
v = '5'+'2'
print(v)
````

> **Common pitfall — you cannot add a number and a string with +**
>
> Mixing types carelessly, like `'5' + 2`, raises a `TypeError`. Python refuses to *guess* whether you meant addition or concatenation — it forces you to be explicit.



````python
u = 5
v = '2'
w = u + v
````

### Explicit conversion

### Convert text into a number

Sometimes a value looks like a number but is stored as text. We can convert suitable text into a number using the built-in functions `int()` or `float()` to perform an **explicit conversion**.

````python
age_text = "20"

age = int(age_text)                           # Convert str to int
print(age)
print(type(age))
print(age + 1)
````

````python
temperature_text = "28.5"

temperature = float(temperature_text)         # Convert str to float
print(temperature)
print(type(temperature))
print(temperature + 1)
````

> **Common pitfall — int() fails on non-numeric text**

- `int()` and `float()`  work on strings — **as long as the string's content really is a number**. If it isn't, Python raises a `ValueError`.


````python
greet = 'Hello'
i = int(greet)   # ValueError: 'Hello' is not a number -- this cell is expected to crash
````

We can also do conversion in the opposite direction: from int or float to text using the built-in `str()` function

````python
age = 20
temperature = 28.5

age_text = str(age)
print(age_text)
print(type(age_text))

temperature_text = str(temperature)
print(temperature_text)
print(type(temperature_text))
````

In contrast, when an expression mixes an `int` and a `float`, Python automatically converts the `int` to a `float` before doing the arithmetic. This is an example of **implicit conversion**.


````python
i = 42      # int
j = 7.0     # float
k = i + j   # i is implicitly converted to 42.0 before the addition
print(k)
print(type(k))
````

## Interacting with the user: `input()`

`input(prompt)` displays `prompt`, waits for the user to type something and press Enter, and returns
whatever they typed **as a string** (always, even if they typed digits).


Let's type in the following code:

````python
name = input ('What is your name ? ')
print('Hello', name)
````

Because `input()` always returns a `str`, if you want to do arithmetic with the answer, you **must** convert it explicitly with `int()` or `float()`. Forgetting this is one of the most common beginner bugs.

````python
# Wrong
age_text = input('What is your age? ')
print('In 5 years you will be', age_text + 5)   # TypeError: str + int
````

````python
# Correct
age_text = input('What is your age? ')
age_int = int(age_text)
print('In 5 years you will be', age_int + 5)
````

You could also write the following code (same result) by using the same variable `age`.

````python
age = input('What is your age? ')
age = int(age)
print('In 5 years you will be', age + 5)
````

## Part 8: Python is a dynamically typed language

- Variable types are checked **while the program runs**, not before.
- You never declare a variable's type (there is no `int x = 5;` like in some other languages). Python figures the type out from the value you assign.
- The *same* variable name can hold an integer at one point in the program and a string later on; Python does not stop you.


````python
z = 5          # z refers to an int
print(type(z))
z = "five"     # now z refers to a str -- perfectly legal, no error
print(type(z))
````

## Part 9: Passing multiple arguments to `print()`

You can pass several values to `print()`, separated by commas. Python automatically inserts a space between each item and converts non-string values (numbers, booleans, ...) to text for you.


````python
name = 'Exaucé'
age = 24

print("My name is", name, "and I am", age, "years old.")
````

### A convenient option: f-strings

An **f-string** lets you place variable values inside a sentence. Put `f` before the quotation mark and place each variable name inside curly brackets `{}`.

```python
f"text {variable} text"
```

````python
name = "Exaucé"
age = 24

print(f"My name is {name} and I am {age} years old.")
````

````python
height = 1.74

print(f"I am {height} meters tall.")
````
